# Numpy Essentials
NumPy is a Python library used for working with arrays. NumPy aims to provide an array object that is up to 50x faster than traditional Python lists.

The array object in NumPy is called ndarray, it provides a lot of supporting funstions that make working with ndarray very easy.

```python
# Computing element-wise operations

import numpy as np

# create two arrays
a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])

# element-wise operation
result = a + b
print(result)
```

```python
# reverse np array
reverseB  = np.flip(b)

print(a + reverseB)
```

```python
# Broadcasting

# create an array and scalar
a = np.array([1, 2, 3, 4])
scalar = 10

# scalar multiplication
print(a * scalar)
```

```python
print(scalar * a)

# print(a * scalar) or print(scalar * a)
# Both are same.
```

```python
# Generate 5 reandom numbers from a uniform distribution
randomNumbers = np.random.uniform(size=5)
print(randomNumbers)
```

```python
# create two 2-D matrices
A = np.array([ [1, 2], [3, 4] ])
B = np.array([ [5, 6], [7, 8]] )

print(A)
```

```python
print(B)
```

```python
matrix multiplication
print( np.dot(A, B) )
```
