# Pandas Essentials

```python
import pandas as pd
print(pd.__version__)
```

```python
# creating a series
data=[x for x in range(1, 6)]
series = pd.Series(data)
print(series)
```

```python
# creating a dataframe
data={'Name':['Alice', 'Bob', 'David', 'Morrison'],
      'Age':[25, 30, 35, 40],
      'City':['New York', 'Los Angeles', 'Chicago', 'Houston']}
df=pd.DataFrame(data)
df
```

```python
df.info()
```

```python
df.describe()
```

```python
# indexing & slicing
# select a single column by position
print(df.iloc[:,0])
```

```python
# select rows & column by position
print(df.loc[1:3, ['Name', 'City']])
```

```python
# select rows & columns by position
df.iloc[0:3, 0:2]
```

```python
# select a single column by label
df['Name']
```

```python
# include null data
nullData={'Name':['Alice', None, 'David', 'Morrison'],
      'Age':[25, 30, None, 40],
      'City':['New York', 'Los Angeles', 'Chicago', 'Houston']}
nullDf=pd.DataFrame(nullData)
nullDf
```

```python
# Missing Data Handling
# Drop rows with missing values
nullDf.dropna()
```

```python
# fill missing value with a specified value
nullDf.fillna(0)
```
