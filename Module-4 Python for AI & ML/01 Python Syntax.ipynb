```python
# class defination
class Person:
    def __init__(self, name, age):
        self.name=name
        self.age=age
    def greet(self):
        print("Hello, my name is ", self.name)
        print("My age is ", self.age)

# object creation
person1=Person("Alice", 30)
person1.greet()
```
